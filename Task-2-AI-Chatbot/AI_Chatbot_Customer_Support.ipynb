```python
customer_queries = {
    "hello": "Hello! Welcome to our AI Customer Support. How may I assist you today?",
    "hi": "Hi! How can I help you today?",
    "order status": "Please provide your Order ID to check the current status of your order.",
    "refund": "Refund requests are processed within 5–7 business days after approval.",
    "cancel order": "Your order can be cancelled before it has been shipped.",
    "delivery": "Orders are usually delivered within 3–5 business days.",
    "payment": "We accept UPI, Credit Cards, Debit Cards, Net Banking, and Cash on Delivery.",
    "contact": "You can contact our support team at support@example.com.",
    "okay": "Yes! Glad I could help. Is there anything else I can assist you with?",
    "ok": "Yes! Glad I could help. Is there anything else I can assist you with?",
    "yes": "Great! Please let me know your next query.",
    "no": "Alright! Feel free to reach out anytime if you need assistance.",
    "thanks": "You're welcome! Happy to assist you.",
    "thank you": "You're most welcome! Have a great day!",
    "bye": "Thank you for using our AI Customer Support Chatbot. Have a wonderful day!",
    "good morning": "Good morning! How can I help you today?",
    "good afternoon": "Good afternoon! How may I assist you?",
    "good evening": "Good evening! How can I help you?",
    "who are you": "I'm an AI Customer Support Chatbot designed to assist you with your queries.",
    "help": "I'm here to help. You can ask me about orders, refunds, delivery, payments, or cancellations.",
    "great": "I'm glad you found my assistance helpful!",
    "exchange": "Products can be exchanged within 7 days.",
"track order": "Please enter your Order ID to track your shipment.",
"shipping charges": "Free shipping is available on orders above ₹499.",
"working hours": "Customer support is available from 9 AM to 6 PM.",
"location": "We provide services across India.",
}
```

```python
def chatbot():
    print("AI Customer Support Chatbot")
    print("Type 'bye' to exit.\n")

    running = True

    while running:
      user = input("You: ").lower()

      if "hello" in user or "hi" in user:
            print("Bot:", customer_queries["hello"])

      elif "refund" in user:
            print("Bot:", customer_queries["refund"])

      elif "cancel" in user:
            print("Bot:", customer_queries["cancel order"])

      elif "delivery" in user:
            print("Bot:", customer_queries["delivery"])

      elif "payment" in user:
            print("Bot:", customer_queries["payment"])

      elif "order" in user:
            print("Bot:", customer_queries["order status"])

      elif "contact" in user:
            print("Bot:", customer_queries["contact"])

      elif "who are you" in user:
            print("Bot:", customer_queries["who are you"])

      elif "okay" in user or "ok" in user:
            print("Bot:", customer_queries["okay"])

      elif "thanks" in user or "thank you" in user:
            print("Bot:", customer_queries["thanks"])

      elif "bye" in user:
            print("Bot:", customer_queries["bye"])
            running = False

      else:
            print("Bot: Sorry, I couldn't understand your query. Please try another question.")

chatbot()
```
